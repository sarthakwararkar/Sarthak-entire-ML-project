```python
numeric_features = [feature for feature in df.columns if df[feature].dtype !='O']
categorical_features = [feature for feature in df.columns if df[feature].dtype =='O']

print('we have {} num feat :{}'.format(len(numeric_features) ,numeric_features))
print('we have {} cat feat :{}'.format(len(categorical_features) ,categorical_features))
```

```python
df['total_score']= df['math_score'] + df['reading_score'] + df['writing_score']
df['avg_score'] = df['total_score'] /3

df.head()
```

```python
fig , axs= plt.subplots(1 ,2  , figsize=(15 ,7)) 
plt.subplot(121)
sns.histplot(data=df , x ='avg_score' , bins=30 , kde =True , color='g')
plt.subplot(122)
sns.histplot(data=df , x='avg_score' , kde =True  , hue='gender')
plt.show()
```
